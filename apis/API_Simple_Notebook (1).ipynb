## Example 1: Your First API Call

Let's get a single blog post from a free test API called JSONPlaceholder.

```python
# Make a GET request
url = "https://jsonplaceholder.typicode.com/posts/1"
response = requests.get(url)

# Check if it worked
print(f"Status Code: {response.status_code}")
print(f"Success: {response.status_code == 200}")

# See the response
print("\nResponse:")
print(response.text)
```

## Example 2: Parsing JSON Data

Let's convert that response into a Python dictionary so we can work with it:

```python
# Parse the JSON response
data = response.json()

# Now we can access fields like a dictionary
print(f"Post ID: {data['id']}")
print(f"User ID: {data['userId']}")
print(f"Title: {data['title']}")
print(f"\nBody: {data['body']}")
```

## Example 3: Getting Multiple Items

Let's get several posts at once:

```python
# Get all posts
url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)
posts = response.json()

print(f"Total posts received: {len(posts)}")
print("\nFirst 5 post titles:")
for post in posts[:5]:
    print(f"  {post['id']}: {post['title']}")
```

## Example 4: Using Parameters to Filter

We can add parameters to customize what data we get back:

```python
# Get only posts from user 1
url = "https://jsonplaceholder.typicode.com/posts"
params = {"userId": 1}

response = requests.get(url, params=params)
user_posts = response.json()

print(f"User 1 has {len(user_posts)} posts")
print("\nTheir post titles:")
for post in user_posts:
    print(f"  - {post['title']}")
```

## Example 5: Creating Data with POST

Let's create a new blog post:

```python
# Create a new post
url = "https://jsonplaceholder.typicode.com/posts"

new_post = {
    "title": "My First API Post",
    "body": "I just learned how to use APIs!",
    "userId": 1
}

response = requests.post(url, json=new_post)
created_post = response.json()

print(f"Status Code: {response.status_code}")
print(f"Created post with ID: {created_post['id']}")
print(f"\nFull response:")
print(json.dumps(created_post, indent=2))
```

## Example 6: Error Handling

What happens when things go wrong? Let's see:

```python
# Try to get a post that doesn't exist
url = "https://jsonplaceholder.typicode.com/posts/99999"
response = requests.get(url)

print(f"Status Code: {response.status_code}")

if response.status_code == 200:
    print("Success!")
    print(response.json())
elif response.status_code == 404:
    print("Not found - this post doesn't exist")
else:
    print(f"Something went wrong: {response.status_code}")
```

## Example 7: Proper Error Handling with Try-Except

Here's how to handle errors properly:

```python
def get_post(post_id):
    """
    Get a post by ID with proper error handling.
    """
    url = f"https://jsonplaceholder.typicode.com/posts/{post_id}"

    try:
        response = requests.get(url, timeout=10)
        response.raise_for_status()  # Raises an error for bad status codes
        return response.json()

    except requests.exceptions.Timeout:
        print("Request timed out")
        return None

    except requests.exceptions.HTTPError as e:
        print(f"HTTP Error: {e}")
        return None

    except requests.exceptions.RequestException as e:
        print(f"Error: {e}")
        return None

# Test it with a valid post
print("Getting post 1:")
post = get_post(1)
if post:
    print(f"  Title: {post['title']}")

# Test it with an invalid post
print("\nGetting post 99999:")
post = get_post(99999)
if not post:
    print("  Failed as expected")
```

## Example 9: Building a Simple Wrapper Function

Let's create a reusable function for working with posts:

```python
class SimpleBlogAPI:
    """
    A simple wrapper for the JSONPlaceholder API.
    """

    def __init__(self):
        self.base_url = "https://jsonplaceholder.typicode.com"

    def get_post(self, post_id):
        """Get a single post by ID."""
        url = f"{self.base_url}/posts/{post_id}"
        response = requests.get(url)
        if response.status_code == 200:
            return response.json()
        return None

    def get_all_posts(self):
        """Get all posts."""
        url = f"{self.base_url}/posts"
        response = requests.get(url)
        if response.status_code == 200:
            return response.json()
        return []

    def get_user_posts(self, user_id):
        """Get all posts by a specific user."""
        url = f"{self.base_url}/posts"
        params = {"userId": user_id}
        response = requests.get(url, params=params)
        if response.status_code == 200:
            return response.json()
        return []

# Use the wrapper
api = SimpleBlogAPI()

# Get a single post
post = api.get_post(1)
print(f"Post 1 title: {post['title']}")

# Get posts by user 2
user_posts = api.get_user_posts(2)
print(f"\nUser 2 has {len(user_posts)} posts")

# Get all posts
all_posts = api.get_all_posts()
print(f"\nTotal posts available: {len(all_posts)}")
```
