A **regular expression** (aka **regex**) is a sequence of characters that define a search pattern, mainly used for pattern matching with strings. Regex is a *universal* string manipulation library which is used in many differet software packages. For example, the equivalent in R is called `stringr`. 

There are a few main uses of regular expressions:
1. **Validating data**, like e-mail addresses, ZIP codes, web page addresses, social security numbers etc. If this is what you need to do, you'll rarely have to create your own regular expressions. For common items like these, visit websites like:  
    - https://regex101.com
    - https://www.regexlib.com
    - https://www.regular-expressions.info
2. **Extracting data from text**, sometimes known as scraping
3. **Cleaning data**
4. **Transforming data into other formats**, for example transforming data that was collected as tab-separated or space-separated into comma-separated values


There are two things you need to learn to workwith regex in Python: (1) the regex syntax used for pattern matching and (2) the Python functions which use regex patterns to search strings. Typically, regex patterns consist of a combination of alphanumeric characters and also special characters, or metacharacters: `. \ | ( ) [ { $ * + ?`. To match meta-characters, you need to escape them with a `\`. We will start by learning about some of the more useful functions below.

# More on regex 

Let's say we want to return all the lines that end in the "or" sound

We can see that the "or" sound can be spelled "oor" (door) or "ore" (Lenore, before) so we add that pattern.  
Also, to search from the beginning of a line we use the "^" symbol  
To search to the end of a line we use "$"  
(Remember that if we want to search for an actual dollar sign, we need to escape it with "\")


````python
pattern = r'(oor|ore)$'
matches = []
for line in lines:
    if re.search(pattern, line, re.IGNORECASE):
        matches.append(line.strip())

print("Lines ending with 'oor' or 'ore':")
for match in matches:
    print(match)

````

# Bag of words

The first step in creating a bag of words is to to tokenize the text  
This involves splitting the text on white space into individual components  

We will start tokenization with nltk which has a word_tokenize method  
It uses the punkt object to tokenize  

````python
# nltk tokenization
# scikit-learn will tokenize but not stem

# Tokenization without stemming
tokens = word_tokenize(poem)
print("Total tokens:", len(tokens))
print("First 20 tokens:", tokens[:20])
````

# Punctuation 
The punctuation is in the token list but generally we want to exclude it.  

## Options  
We have a couple of options:
1) We can remove the punctuation after the fact
- Python's `string` library has a `punctuation` attribute that stores a list of punctuation marks (`string.punctuation`).
2) We can use regular expressions to replace punctuation _before_ tokenizing
- We will again use `string.punctuation` but we could as easily use the `regex` package and the `[[:punct:]]` syntax.  

````python
# Option 1: Filter after the fact
import string
filtered_tokens = [word for word in tokens if word not in string.punctuation]
print("First 20 tokens:", filtered_tokens[:20])
print(f"Total number of tokens: {len(filtered_tokens)}")
````

````python
# Option 2: regex to replace the punctuation
# re.escape will add the escape character to string.punctuation objects so we do not have to manually do so
# for example, . or - need to be escaped 
# The square brackets will look for _any_ of the punctuation marks
punc_pattern = "[" + re.escape(string.punctuation) + "]+"
poem2 = re.sub(punc_pattern,"", poem)
#print(poem2)
tokens2 = word_tokenize(poem2)
print("First 20 tokens:", tokens2[:20])
print(f"Total number of tokens: {len(filtered_tokens)}")
````

# Stemming  
In many cases, we may want to stem our tokens  
Stemming involves removing the suffixes of words  
It is a simple way to reduce roots to (hopefully) their root values  

````python
# Tokenization with stemming
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(token) for token in tokens2]
print("Total stemmed tokens:", len(stemmed_tokens))
print("First 20 stemmed tokens:", stemmed_tokens[:20])
````

# scikit-learn

Now we will see what `scikit-learn` can do.  
It does have a built in tokenizer and it will create n-grams (tokens of more than one word)  

But it will not stem so we would have to stem before hand  

````python
# Bag of words without stemming
# We now know that the code below will initiate an object of the CountVectorizer class
# We will use methods of the class in order to perform operations on our data
vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform([poem])
?CountVectorizer.fit_transform
````

````python
# We will be using the same code to show the top words so we will create a function

def show_top_words(features, counts):
    word_counts = list(zip(features, counts))
    word_counts.sort(key=lambda x: x[1], reverse=True)
    print("Top 10 most common words:")
    for word, count in word_counts[:10]:
        print(f"\t{word}: {count}")
````

`bow_matrix` is a new bag of words matrix object.  
From the help we can see that it returns an array with the number of documents as rows and number of features as columns  

In the next block of code, we will explore how to look at the data  
`vectorizer.get_feature_names_out()` will return the actual feature names  
The array itself does not have the feature names  

To get the token counts, we need to convert the sparse matrix scikit creates into an array  



````python
print("Bag of words matrix type:", type(bow_matrix))
print("Shape:", bow_matrix.shape)
print("Non-zero elements:", bow_matrix.nnz)
features = vectorizer.get_feature_names_out()
counts = bow_matrix.toarray()[0]
show_top_words(features,counts)
````

````python
# Define custom tokenizer with stemming for CountVectorizer
def stemmed_tokenizer(text):
    tokens = word_tokenize(text)
    stemmer = PorterStemmer()
    return [stemmer.stem(token) for token in tokens]
````

````python
# Bag of words with stemming
vectorizer_stemmed = CountVectorizer(tokenizer=stemmed_tokenizer)
bow_stemmed_matrix = vectorizer_stemmed.fit_transform([poem2])
print("Stemmed bag of words matrix type:", type(bow_stemmed_matrix))
print("Shape:", bow_stemmed_matrix.shape)
print("Non-zero elements:", bow_stemmed_matrix.nnz)
counts_stemmed = bow_stemmed_matrix.toarray()[0]
features_stemmed = vectorizer_stemmed.get_feature_names_out()
show_top_words(features_stemmed,counts_stemmed)
````
