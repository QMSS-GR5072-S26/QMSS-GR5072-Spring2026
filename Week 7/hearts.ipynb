````python
import numpy as np
rng = np.random.default_rng()
def round_score():
    """
    Function to score a round of hearts_game
    
    Returns:
        list of 4 values adding to 26
    """
    #round_score = [0]*4
    round_score = []
    # 13 hearts to be distributed
    remaining = 13
    for player in range(4):
        if player<3:
            score = rng.integers(0,remaining, endpoint=True)
            round_score.append(score)
            remaining -= score
        else:
            round_score.append(remaining)
    # One person gets the Queen of Spades (13 points)
    # Python starts indexing at 0 so we have to start there rather than 1
    # Also, rng.integers is minimum inclusive and maximum inclusive with endpoint=True
    queen = rng.integers(0,3, endpoint=True)
    round_score[queen] += 13
    return round_score
````
